```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
```

```python
business_ids = [f"B{i:03d}" for i in range(1, 501)]
business_ids[:5], business_ids[-5:]
```

```python
owner_ids = [f"O{i:03d}" for i in range(1, 501)]
owner_ids[:5], owner_ids[-5:]
```

```python
business_types = (
    ["Retail Shop"] * 100 +
    ["Food Vendor"] * 80 +
    ["Transport Worker"] * 60 +
    ["Artisan"] * 50 +
    ["Freelancer"] * 50 +
    ["Small Producer"] * 60 +
    ["Home Business"] * 50 +
    ["Micro Service Business"] * 50
)

len(business_types)
```

```python
business_types = np.random.permutation(business_types)

business_types[:10]
```

```python
pd.Series(business_types).value_counts()
```

```python
business_df = pd.DataFrame({
    "business_id": business_ids,
    "owner_id": owner_ids,
    "business_type": business_types
})

business_df.head()
```

```python
environment_types = (
    ["Urban"] * 175 +
    ["Semi-Urban"] * 125 +
    ["Rural"] * 125 +
    ["Peri-Urban"] * 75
)

environment_types = np.random.permutation(environment_types)

pd.Series(environment_types).value_counts()
```

```python
business_df["environment_type"] = environment_types

business_df.head()
```

```python
locations = [
    "Delhi",
    "Mumbai",
    "Jaipur",
    "Lucknow",
    "Indore",
    "Kochi",
    "Guwahati",
    "Bengaluru",
    "Ahmedabad",
    "Patna"
]

business_df["location"] = np.random.choice(
    locations,
    size=len(business_df)
)

business_df.head()
```

```python
months = pd.date_range(
    start="2024-01-01",
    periods=24,
    freq="MS"
)

months
```

```python
monthly_df = business_df.merge(
    pd.DataFrame({"month": months}),
    how="cross"
)

monthly_df.shape
```

```python
base_income = {
    "Retail Shop": 45000,
    "Food Vendor": 30000,
    "Transport Worker": 28000,
    "Artisan": 32000,
    "Freelancer": 40000,
    "Small Producer": 35000,
    "Home Business": 25000,
    "Micro Service Business": 30000
}
```

```python
expense_ratio = {
    "Retail Shop": 0.72,
    "Food Vendor": 0.65,
    "Transport Worker": 0.55,
    "Artisan": 0.60,
    "Freelancer": 0.40,
    "Small Producer": 0.68,
    "Home Business": 0.50,
    "Micro Service Business": 0.55
}
```

```python
income_variation = np.random.normal(
    loc=1.0,
    scale=0.10,
    size=len(monthly_df)
)
monthly_df["monthly_income"] = (
    monthly_df["business_type"]
    .map(base_income)
    * income_variation
).round(2)
monthly_df["monthly_income"].head()
```

```python
expense_variation = np.random.normal(
    loc=1.0,
    scale=0.05,
    size=len(monthly_df)
)
monthly_df["monthly_expenses"] = (
    monthly_df["monthly_income"]
    * monthly_df["business_type"].map(expense_ratio)
    * expense_variation
).round(2)
monthly_df["monthly_expenses"].head()
```

```python
monthly_df["transaction_count"] = np.maximum(
    1,
    np.round(
        monthly_df["monthly_income"] / 500
        + np.random.normal(0, 8, len(monthly_df))
    )
).astype(int)

monthly_df["transaction_count"].head()
```

```python
digital_payment_base = {
    "Retail Shop": 65,
    "Food Vendor": 45,
    "Transport Worker": 50,
    "Artisan": 40,
    "Freelancer": 85,
    "Small Producer": 35,
    "Home Business": 55,
    "Micro Service Business": 60
}

digital_payment_variation = np.random.normal(
    loc=0,
    scale=8,
    size=len(monthly_df)
)

monthly_df["digital_payment_pct"] = (
    monthly_df["business_type"].map(digital_payment_base)
    + digital_payment_variation
).clip(0, 100).round(2)

monthly_df["digital_payment_pct"].head()
```

```python
supplier_payment_ratio = {
    "Retail Shop": 0.45,
    "Food Vendor": 0.35,
    "Transport Worker": 0.15,
    "Artisan": 0.30,
    "Freelancer": 0.10,
    "Small Producer": 0.40,
    "Home Business": 0.25,
    "Micro Service Business": 0.20
}

supplier_payment_variation = np.random.normal(
    loc=1.0,
    scale=0.08,
    size=len(monthly_df)
)

monthly_df["supplier_payment_amount"] = (
    monthly_df["monthly_expenses"]
    * monthly_df["business_type"].map(supplier_payment_ratio)
    * supplier_payment_variation
).clip(lower=0).round(2)

monthly_df["supplier_payment_amount"].head()
```

```python
utility_ratio = {
    "Retail Shop": 0.08,
    "Food Vendor": 0.07,
    "Transport Worker": 0.04,
    "Artisan": 0.06,
    "Freelancer": 0.03,
    "Small Producer": 0.08,
    "Home Business": 0.05,
    "Micro Service Business": 0.05
}

utility_variation = np.random.normal(
    loc=1.0,
    scale=0.10,
    size=len(monthly_df)
)

monthly_df["utility_payment_amount"] = (
    monthly_df["monthly_expenses"]
    * monthly_df["business_type"].map(utility_ratio)
    * utility_variation
).clip(lower=0).round(2)

monthly_df["utility_payment_amount"].head()
```

```python
rent_base = {
    "Urban": 9000,
    "Semi-Urban": 6500,
    "Rural": 4000,
    "Peri-Urban": 5500
}

rent_variation = np.random.normal(
    loc=1.0,
    scale=0.08,
    size=len(monthly_df)
)

monthly_df["rent_payment"] = (
    monthly_df["environment_type"].map(rent_base)
    * rent_variation
).clip(lower=0).round(2)

monthly_df["rent_payment"].head()
```

```python
cash_before_savings = (
    monthly_df["monthly_income"]
    - monthly_df["monthly_expenses"]
    - monthly_df["rent_payment"]
)

savings_rate = np.random.uniform(
    0.05,
    0.25,
    size=len(monthly_df)
)

monthly_df["savings_amount"] = (
    cash_before_savings * savings_rate
).clip(lower=0).round(2)

monthly_df["savings_amount"].head()
```

```python
monthly_df["net_cash_flow"] = (
    monthly_df["monthly_income"]
    - monthly_df["monthly_expenses"]
    - monthly_df["rent_payment"]
    - monthly_df["utility_payment_amount"]
).round(2)

monthly_df["net_cash_flow"].head()
```

```python
payment_discipline_base = {
    "Retail Shop": 82,
    "Food Vendor": 78,
    "Transport Worker": 75,
    "Artisan": 80,
    "Freelancer": 85,
    "Small Producer": 76,
    "Home Business": 81,
    "Micro Service Business": 83
}

payment_discipline_variation = np.random.normal(
    loc=0,
    scale=7,
    size=len(monthly_df)
)

monthly_df["payment_discipline_pct"] = (
    monthly_df["business_type"].map(payment_discipline_base)
    + payment_discipline_variation
).clip(0, 100).round(2)

monthly_df["payment_discipline_pct"].head()
```

```python
supplier_reliability_base = {
    "Retail Shop": 78,
    "Food Vendor": 75,
    "Transport Worker": 72,
    "Artisan": 76,
    "Freelancer": 82,
    "Small Producer": 74,
    "Home Business": 77,
    "Micro Service Business": 80
}

supplier_reliability_variation = np.random.normal(
    loc=0,
    scale=8,
    size=len(monthly_df)
)

monthly_df["supplier_reliability_pct"] = (
    monthly_df["business_type"].map(supplier_reliability_base)
    + supplier_reliability_variation
).clip(0, 100).round(2)

monthly_df["supplier_reliability_pct"].head()
```

```python
continuity_base = {
    "Retail Shop": 88,
    "Food Vendor": 82,
    "Transport Worker": 80,
    "Artisan": 84,
    "Freelancer": 90,
    "Small Producer": 78,
    "Home Business": 86,
    "Micro Service Business": 88
}

continuity_variation = np.random.normal(
    loc=0,
    scale=6,
    size=len(monthly_df)
)

monthly_df["business_continuity_pct"] = (
    monthly_df["business_type"].map(continuity_base)
    + continuity_variation
).clip(0, 100).round(2)

monthly_df["business_continuity_pct"].head()
```

```python
demand_stability_base = {
    "Retail Shop": 78,
    "Food Vendor": 72,
    "Transport Worker": 68,
    "Artisan": 70,
    "Freelancer": 80,
    "Small Producer": 74,
    "Home Business": 73,
    "Micro Service Business": 79
}

demand_stability_variation = np.random.normal(
    loc=0,
    scale=8,
    size=len(monthly_df)
)

monthly_df["demand_stability_pct"] = (
    monthly_df["business_type"].map(demand_stability_base)
    + demand_stability_variation
).clip(0, 100).round(2)

monthly_df["demand_stability_pct"].head()
```

```python
seasonal_factors = {
    1: 0.95,
    2: 0.97,
    3: 1.00,
    4: 1.02,
    5: 1.05,
    6: 1.08,
    7: 1.03,
    8: 1.00,
    9: 0.98,
    10: 1.10,
    11: 1.15,
    12: 1.20
}

monthly_df["seasonal_factor"] = monthly_df["month"].dt.month.map(
    seasonal_factors
)

monthly_df[["month", "seasonal_factor"]].head(15)
```

```python
monthly_df["seasonally_adjusted_income"] = (
    monthly_df["monthly_income"]
    * monthly_df["seasonal_factor"]
).round(2)

monthly_df[[
    "month",
    "monthly_income",
    "seasonal_factor",
    "seasonally_adjusted_income"
]].head(12)
```

```python
shock_types = [
    "None",
    "Demand Drop",
    "Supply Disruption",
    "Operating Cost Increase"
]

shock_probabilities = [0.82, 0.07, 0.06, 0.05]

monthly_df["shock_type"] = np.random.choice(
    shock_types,
    size=len(monthly_df),
    p=shock_probabilities
)

shock_factor_map = {
    "None": 1.00,
    "Demand Drop": 0.75,
    "Supply Disruption": 0.85,
    "Operating Cost Increase": 0.90
}

monthly_df["shock_factor"] = monthly_df["shock_type"].map(
    shock_factor_map
)

monthly_df["shock_type"].value_counts()
```

```python
monthly_df["adjusted_income"] = (
    monthly_df["seasonally_adjusted_income"]
    * monthly_df["shock_factor"]
).round(2)

monthly_df["adjusted_expenses"] = (
    monthly_df["monthly_expenses"]
    * np.where(
        monthly_df["shock_type"].eq("Operating Cost Increase"),
        1.20,
        np.where(
            monthly_df["shock_type"].eq("Supply Disruption"),
            1.10,
            1.00
        )
    )
).round(2)

monthly_df["adjusted_cash_flow"] = (
    monthly_df["adjusted_income"]
    - monthly_df["adjusted_expenses"]
    - monthly_df["rent_payment"]
    - monthly_df["utility_payment_amount"]
).round(2)

monthly_df[
    [
        "shock_type",
        "adjusted_income",
        "adjusted_expenses",
        "adjusted_cash_flow"
    ]
].head(10)
```

```python
# Cash-flow stability
monthly_df["cash_flow_margin_pct"] = (
    monthly_df["adjusted_cash_flow"]
    / monthly_df["adjusted_income"].replace(0, np.nan)
    * 100
).clip(-100, 100).fillna(0).round(2)

# Financial pressure
monthly_df["expense_burden_pct"] = (
    monthly_df["adjusted_expenses"]
    / monthly_df["adjusted_income"].replace(0, np.nan)
    * 100
).clip(0, 200).fillna(0).round(2)

# Overall operational stability
monthly_df["operational_stability_pct"] = (
    monthly_df[
        [
            "payment_discipline_pct",
            "supplier_reliability_pct",
            "business_continuity_pct",
            "demand_stability_pct"
        ]
    ].mean(axis=1)
).round(2)

# Financial health
monthly_df["financial_health_pct"] = (
    monthly_df["cash_flow_margin_pct"].clip(0, 100) * 0.45
    + (100 - monthly_df["expense_burden_pct"].clip(0, 100)) * 0.30
    + monthly_df["payment_discipline_pct"] * 0.25
).clip(0, 100).round(2)

# Resilience
monthly_df["resilience_pct"] = (
    monthly_df[
        [
            "business_continuity_pct",
            "demand_stability_pct",
            "supplier_reliability_pct"
        ]
    ].mean(axis=1)
).round(2)

monthly_df[
    [
        "cash_flow_margin_pct",
        "expense_burden_pct",
        "operational_stability_pct",
        "financial_health_pct",
        "resilience_pct"
    ]
].head()
```

```python
monthly_df["economic_reputation_index"] = (
    monthly_df["financial_health_pct"] * 0.35
    + monthly_df["operational_stability_pct"] * 0.25
    + monthly_df["resilience_pct"] * 0.20
    + monthly_df["payment_discipline_pct"] * 0.20
).clip(0, 100).round(2)

monthly_df[
    [
        "business_id",
        "month",
        "financial_health_pct",
        "operational_stability_pct",
        "resilience_pct",
        "economic_reputation_index"
    ]
].head()
```

```python
print("Dataset shape:", monthly_df.shape)
print("Duplicate rows:", monthly_df.duplicated().sum())
print("Missing values:", monthly_df.isna().sum().sum())
print("Unique businesses:", monthly_df["business_id"].nunique())
print("Unique months:", monthly_df["month"].nunique())

print("\nBusiness type distribution:")
print(monthly_df["business_type"].value_counts())

print("\nShock distribution:")
print(monthly_df["shock_type"].value_counts())

print("\nReputation statistics:")
print(monthly_df["economic_reputation_index"].describe())
```
