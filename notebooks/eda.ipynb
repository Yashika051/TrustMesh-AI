# Exploratory Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

df = pd.read_csv(
    "../data/trustmesh_economic_data.csv",
    parse_dates=["month"],
    keep_default_na=False
)

print(f"Dataset shape: {df.shape}")
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Missing values: {df.isna().sum().sum()}")

df.head()
```

```python

print("Shape:", df.shape)
print("\nBusiness types:")
print(df["business_type"].value_counts())

print("\nEnvironment types:")
print(df["environment_type"].value_counts())

print("\nLocations:")
print(df["location"].value_counts())

print("\nDate range:")
print(df["month"].min(), "to", df["month"].max())

df.describe().T
```

```python

fig = px.bar(
    df["business_type"].value_counts().reset_index(),
    x="business_type",
    y="count",
    title="Business Type Distribution",
    labels={"business_type": "Business Type", "count": "Number of Observations"}
)

fig.show()
```

```python
df["location"].value_counts()
```

```python


financial_by_type = (
    df.groupby("business_type")
      .agg(
          avg_income=("adjusted_income", "mean"),
          avg_expenses=("adjusted_expenses", "mean"),
          avg_cash_flow=("adjusted_cash_flow", "mean"),
          avg_transactions=("transaction_count", "mean"),
          avg_digital_payment=("digital_payment_pct", "mean"),
          avg_expense_burden=("expense_burden_pct", "mean")
      )
      .round(2)
      .sort_values("avg_cash_flow", ascending=False)
)

display(financial_by_type)


for metric, title, y_label in [
    ("avg_income", "Average Adjusted Income by Business Type", "Average Income"),
    ("avg_cash_flow", "Average Adjusted Cash Flow by Business Type", "Average Cash Flow"),
    ("avg_expense_burden", "Average Expense Burden by Business Type", "Expense Burden (%)")
]:
    fig = px.bar(
        financial_by_type.reset_index(),
        x="business_type",
        y=metric,
        title=title,
        labels={
            "business_type": "Business Type",
            metric: y_label
        }
    )
    fig.show()
```

```python


reputation_by_type = (
    df.groupby("business_type")
      .agg(
          financial_health=("financial_health_pct", "mean"),
          operational_stability=("operational_stability_pct", "mean"),
          resilience=("resilience_pct", "mean"),
          reputation_index=("economic_reputation_index", "mean")
      )
      .round(2)
      .sort_values("reputation_index", ascending=False)
)

display(reputation_by_type)


fig = px.bar(
    reputation_by_type.reset_index(),
    x="business_type",
    y="reputation_index",
    title="Average Economic Reputation Index by Business Type",
    labels={
        "business_type": "Business Type",
        "reputation_index": "Economic Reputation Index"
    }
)

fig.show()


reputation_long = (
    reputation_by_type
    .reset_index()
    .melt(
        id_vars="business_type",
        var_name="metric",
        value_name="score"
    )
)

fig = px.bar(
    reputation_long,
    x="business_type",
    y="score",
    color="metric",
    barmode="group",
    title="Reputation Dimensions by Business Type",
    labels={
        "business_type": "Business Type",
        "score": "Score",
        "metric": "Dimension"
    }
)

fig.show()
```

```python


shock_analysis = (
    df.groupby("shock_type")
      .agg(
          observations=("business_id", "size"),
          avg_income=("adjusted_income", "mean"),
          avg_expenses=("adjusted_expenses", "mean"),
          avg_cash_flow=("adjusted_cash_flow", "mean"),
          avg_financial_health=("financial_health_pct", "mean"),
          avg_resilience=("resilience_pct", "mean"),
          avg_reputation=("economic_reputation_index", "mean")
      )
      .round(2)
      .sort_values("avg_reputation", ascending=False)
)

display(shock_analysis)


fig = px.bar(
    shock_analysis.reset_index(),
    x="shock_type",
    y="avg_cash_flow",
    title="Average Adjusted Cash Flow by Economic Shock",
    labels={
        "shock_type": "Economic Shock",
        "avg_cash_flow": "Average Adjusted Cash Flow"
    }
)

fig.show()


fig = px.bar(
    shock_analysis.reset_index(),
    x="shock_type",
    y="avg_reputation",
    title="Average Economic Reputation by Economic Shock",
    labels={
        "shock_type": "Economic Shock",
        "avg_reputation": "Economic Reputation Index"
    }
)

fig.show()
```

```python


monthly_trends = (
    df.groupby("month")
      .agg(
          avg_income=("adjusted_income", "mean"),
          avg_cash_flow=("adjusted_cash_flow", "mean"),
          avg_reputation=("economic_reputation_index", "mean")
      )
      .reset_index()
)

display(monthly_trends.head())
fig = px.line(
    monthly_trends,
    x="month",
    y="avg_income",
    title="Monthly Average Adjusted Income",
    labels={
        "month": "Month",
        "avg_income": "Average Adjusted Income"
    }
)
fig.show()


fig = px.line(
    monthly_trends,
    x="month",
    y="avg_cash_flow",
    title="Monthly Average Adjusted Cash Flow",
    labels={
        "month": "Month",
        "avg_cash_flow": "Average Adjusted Cash Flow"
    }
)
fig.show()


fig = px.line(
    monthly_trends,
    x="month",
    y="avg_reputation",
    title="Monthly Average Economic Reputation",
    labels={
        "month": "Month",
        "avg_reputation": "Economic Reputation Index"
    }
)
fig.show()
```

```python


numeric_cols = df.select_dtypes(include="number").columns

reputation_corr = (
    df[numeric_cols]
    .corr()["economic_reputation_index"]
    .drop("economic_reputation_index")
    .sort_values(ascending=False)
)

display(reputation_corr.to_frame("correlation"))

corr_cols = [
    "adjusted_income",
    "adjusted_expenses",
    "adjusted_cash_flow",
    "digital_payment_pct",
    "payment_discipline_pct",
    "supplier_reliability_pct",
    "business_continuity_pct",
    "demand_stability_pct",
    "cash_flow_margin_pct",
    "expense_burden_pct",
    "financial_health_pct",
    "operational_stability_pct",
    "resilience_pct",
    "economic_reputation_index"
]

fig = px.imshow(
    df[corr_cols].corr().round(2),
    text_auto=True,
    title="Correlation Matrix of TrustMesh Economic Indicators",
    aspect="auto"
)

fig.show()
```

## Key Findings

### 1. Financial Performance Varies Across Business Types

Retail Shops generate the highest average adjusted income at approximately ₹45,351 per month. However, Freelancers generate the highest average adjusted cash flow at approximately ₹17,513, despite having lower average income than Retail Shops.

This difference is largely explained by expense burden. Retail Shops have an average expense burden of approximately 73.42%, compared with only 40.64% for Freelancers.

### 2. Economic Reputation Is Not Determined by Income Alone

Freelancers have the highest average Economic Reputation Index at 75.18, while Retail Shops have a lower average reputation score of 64.44 despite generating the highest average income.

This demonstrates that TrustMesh evaluates broader economic health rather than relying on revenue alone.

### 3. Financial Health Is the Strongest Reputation Signal

Financial health has the strongest correlation with the Economic Reputation Index (r = 0.84), followed by operational stability (r = 0.76), adjusted cash flow (r = 0.74), and cash-flow margin (r = 0.72).

Expense burden has a strong negative relationship with reputation (r = -0.67).

### 4. Economic Shocks Reduce Financial Performance and Reputation

Businesses experiencing economic shocks show substantially weaker financial outcomes than businesses without shocks.

Demand Drop produces the largest average cash-flow impact, with average adjusted cash flow falling to approximately -₹2,181 compared with ₹6,881 for businesses without a shock.

Demand Drop also produces the lowest average financial health (28.46) and Economic Reputation Index (61.13).

### 5. Reputation Changes Over Time

The monthly analysis shows variation in income, cash flow, and Economic Reputation across the 24-month observation period.

The seasonal factors introduced into the dataset create observable changes in economic performance across different months, while economic shocks introduce additional variation.

### 6. Behavioral Signals Contribute to Economic Reputation

Payment discipline, business continuity, supplier reliability, and demand stability all show positive relationships with the Economic Reputation Index.

This supports the idea that TrustMesh can evaluate businesses using a combination of financial and behavioral signals rather than financial performance alone.

### Overall Observation

The exploratory analysis indicates that economic reputation is better represented as a multidimensional measure of financial health, operational stability, resilience, and business behavior rather than a simple measure of business revenue.
